```python
from PIL import Image
from pathlib import Path
import pandas as pd
import cv2
import statistics

from configuration.config import data_path, debug_crop_dir
```

```python
pixels = []
widths = []
hights = []

for img_path in debug_crop_dir.glob("*.jpg"):
    img = Image.open(img_path)
    pixels.append(img.size[0] * img.size[1])
    widths.append(img.size[0])
    hights.append(img.size[1])
```

```python
statistics.mean(pixels) , statistics.mean(widths), statistics.mean(hights)
```

```python
pd.DataFrame({
    "pixels": pixels,
    "widths": widths,
    "hights": hights
}).to_csv(data_path / "debug_crop_stats.csv", index=False)
```
