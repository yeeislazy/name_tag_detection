```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split

from configuration.config import data_path
```

```python
true_list = list((data_path / "true").glob("*.png"))
for i, img_path in enumerate(true_list):
    true_list[i] = img_path.relative_to(data_path) 

false_list = list((data_path / "false").glob("*.jpg")) # + list((data_path / "fake").glob("*.png"))
for i, img_path in enumerate(false_list):
    false_list[i] = img_path.relative_to(data_path)

true_df = pd.DataFrame(
    {
        "image_path": true_list,
        "label": 1
    }
)
false_df = pd.DataFrame(
    {
        "image_path": false_list,
        "label": 0
    }
)
df = pd.concat([true_df, false_df]).reset_index(drop=True)
```

```python
df.info()
```

```python
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42, stratify=df["label"])
train_df.to_csv(data_path / "train.csv", index=False)
test_df.to_csv(data_path / "test.csv", index=False)
```
