```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split

from configuration.config import name_tag_data_dir
```

```python
true_list = list((name_tag_data_dir / "true").glob("*.png")) + list((name_tag_data_dir / "true").glob("*.jpg")) + list((name_tag_data_dir / "fake").glob("*.png")) + list((name_tag_data_dir / "fake").glob("*.jpg"))
for i, img_path in enumerate(true_list):
    true_list[i] = img_path.relative_to(name_tag_data_dir) 

false_list = list((name_tag_data_dir / "false").glob("*.png")) + list((name_tag_data_dir / "false").glob("*.jpg")) # + list((name_tag_data_dir / "fake").glob("*.png")) + list((name_tag_data_dir / "fake").glob("*.jpg"))
for i, img_path in enumerate(false_list):
    false_list[i] = img_path.relative_to(name_tag_data_dir)

true_df = pd.DataFrame(
    {
        "image_path": true_list,
        "label": 1
    }
)
false_df = pd.DataFrame(
    {
        "image_path": false_list,
        "label": 0
    }
)
df = pd.concat([true_df, false_df]).reset_index(drop=True)
```

```python
versions = name_tag_data_dir.glob("v*")
if not versions:
    version = "v1"
else:
    version_numbers = [int(v.name[1:]) for v in versions if v.name[1:].isdigit()]
    if version_numbers:
        version = f"v{max(version_numbers) + 1}"
    else:
        version = "v1"
```

```python
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42, stratify=df["label"])
train_df.to_csv(name_tag_data_dir/ version / "train.csv", index=False)
test_df.to_csv(name_tag_data_dir / version / "test.csv", index=False)
```
